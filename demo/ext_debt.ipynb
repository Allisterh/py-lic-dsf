# Ext_Debt_Data via `load_core`

**Edit instruments / Ext in Excel**, save, then load with `load_core`. Inspect
new-MLT via `external.portfolio` and Ext headlines via `ExternalDebtBook`.

| Layer | Class / API | Role |
|-------|-------------|------|
| Load | `load_core` | Input 4/5 + Ext + Macro + baselines |
| Portfolio | `external.portfolio` (`PVPortfolio`) | New-debt aggregates |
| Book | `ExternalDebtBook` | Existing debt + ST/SDR + Ext_Debt headlines |

See `docs/04-ext-debt-module-design.qmd`.


````python
from __future__ import annotations

from pathlib import Path

import pandas as pd

from lic_dsf.load import load_core, load_stress

REPO_ROOT = Path.cwd().resolve()
if REPO_ROOT.name == "demo":
    REPO_ROOT = REPO_ROOT.parent

WORKBOOK = REPO_ROOT / "data" / "lic-dsf-template-2025-08-12.xlsx"

pd.set_option("display.max_columns", 24)
pd.set_option("display.width", 200)
pd.set_option("display.float_format", lambda x: f"{x:,.4f}")
pd.set_option("display.max_rows", 80)

WORKBOOK
````

## 1. Load via `load_core`

Edit Input 4 / 5 and Ext in Excel, save, then reload. The new-MLT portfolio is
`external.portfolio`.


````python
macro, external, ext_base, pub_base = load_core(WORKBOOK)
book = external
portfolio = external.portfolio
instruments = list(portfolio.instruments)

catalog = pd.DataFrame(
    [
        {
            "name": i.name,
            "grace": i.grace,
            "maturity": i.maturity,
            "interest": i.interest_rate,
            "discount": i.discount_rate,
            "disbursement_sum": sum(i.disbursements),
            "n_years": len(i.disbursements),
        }
        for i in instruments
    ]
)
print(f"{len(instruments)} instruments from {WORKBOOK.name}")
catalog
````

## 2. Spot-check one instrument Output (works today once loaded)

Pick a nonzero line (Eurobond in this template) and show the Ext_Debt-facing Output rows: Interest, Amortization, PV, Stock.

````python
by_name = {i.name: i for i in instruments}
sample_name = "Eurobond" if "Eurobond" in by_name else instruments[0].name
sample = by_name[sample_name]

external = sample.external()
ext_debt_rows = external.loc[
    [
        "Interest",
        "Amortization",
        f"PV of debt   {sample.name}",
        "Stock of new forex debt (in USD)",
    ]
]
print(sample_name, "→ Ext_Debt Output metrics")
ext_debt_rows.iloc[:, :10]
````

grant element % (from `PresentValueInstrument.internal()`, same value Input 4
column I links to). `R407` / `R409` follow from that average and total new
external MLT disbursements.

````python
ge = pd.DataFrame(
    {
        "GE % (R408)": book.grant_element_percent(),
        "disb net of GE (R407)": book.new_disbursements_net_of_ge(),
        "GE value (R409)": book.grant_element_value(),
    }
)
ge.loc[2024:2030]
````

## 9. Residual financing params (Input 7)

Edit Input 7 in Excel, or override programmatically via `resolve_residual_params`.
Decade averages from Ext feed Input 7 **Default** shares/terms; use
`load_stress().residual` for the workbook's resolved terms.


````python
from dataclasses import asdict

from lic_dsf.resfin import ResidualFinancingOverrides, calculate_residual_defaults, resolve_residual_params

defaults = calculate_residual_defaults(book)
resolved = resolve_residual_params(
    defaults,
    ResidualFinancingOverrides(
        external_mlt_share=0.5,
        domestic_mlt_share=0.2,
        avg_interest_rate=6.0,
    )
)
stress = load_stress(WORKBOOK)

pd.DataFrame(
    {
        "default": asdict(defaults),
        "resolved": asdict(resolved),
        "from_workbook": asdict(stress.residual),
    }
)
````
